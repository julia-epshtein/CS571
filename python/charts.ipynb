```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob
import numpy as np
from collections import defaultdict
```

```python
raw_dir = os.path.join('..', 'data', 'processed', 'raw')
measure_dir = os.path.join('..', 'data', 'processed', 'measure')
```

# Data Loading

```python
# Load all measure files
measure_files = sorted(glob.glob(os.path.join(measure_dir, '*_measures_cleaned.csv')))
df_list_measure = []

for file in measure_files:
    year = int(os.path.basename(file)[:4])
    temp_df = pd.read_csv(file)
    temp_df['Year'] = year
    df_list_measure.append(temp_df)

measures_all_years = pd.concat(df_list_measure, ignore_index=True)
```

```python
# Load all raw files
raw_files = sorted(glob.glob(os.path.join(raw_dir, '*_raw_numbers_cleaned.csv')))
df_list_raw = []

for file in raw_files:
    year = int(os.path.basename(file)[:4])
    temp_df = pd.read_csv(file)
    temp_df['Year'] = year
    df_list_raw.append(temp_df)

raw_all_years = pd.concat(df_list_raw, ignore_index=True)
```

```python
# Save measures_all_years and raw_all_years to txt
measures_all_years.to_csv('../data/all_years/measures_all_years.csv', index=False)
raw_all_years.to_csv('../data/all_years/raw_all_years.csv', index=False)
```

# Data Exploration

```python
# Seeing which columns are unique for measures
column_years = defaultdict(list)
year_list = [df['Year'].iloc[0] for df in df_list_measure]  

for df in df_list_measure:
    year = df['Year'].iloc[0]
    for col in df.columns:
        column_years[col].append(year)

# Find all unique year combinations that share columns
year_combinations = defaultdict(list)

# For each column, get the sorted tuple of years it appears in
for col, years in column_years.items():
    year_tuple = tuple(sorted(years))
    year_combinations[year_tuple].append(col)

# Sort the combinations by number of years (from single years to all years)
sorted_combinations = sorted(year_combinations.keys(), 
                            key=lambda x: (len(x), x))

# Print the results
for combo in sorted_combinations:
    years_str = ", ".join(map(str, combo)) if len(combo) > 1 else f"{combo[0]} only"
    print(f"\nColumns shared by {years_str}:")
    for col in sorted(year_combinations[combo]):
        print(f"  - {col}")
    print(f"Total: {len(year_combinations[combo])} columns")
```

```python
# Seeing which columns are unique for raw
column_years = defaultdict(list)
year_list = [df['Year'].iloc[0] for df in df_list_raw]  # Get list of years

for df in df_list_raw:
    year = df['Year'].iloc[0]
    for col in df.columns:
        column_years[col].append(year)

# Find all unique year combinations that share columns
year_combinations = defaultdict(list)

# For each column, get the sorted tuple of years it appears in
for col, years in column_years.items():
    year_tuple = tuple(sorted(years))
    year_combinations[year_tuple].append(col)

# Sort the combinations by number of years (from single years to all years)
sorted_combinations = sorted(year_combinations.keys(), 
                            key=lambda x: (len(x), x))

# Print results
for combo in sorted_combinations:
    years_str = ", ".join(map(str, combo)) if len(combo) > 1 else f"{combo[0]} only"
    print(f"\nColumns shared by {years_str}:")
    for col in sorted(year_combinations[combo]):
        print(f"  - {col}")
    print(f"Total: {len(year_combinations[combo])} columns")
```

```python
# Timeline For Total Imprisonment Across Strike Type
plt.figure(figsize=(14, 8))

# Standard offense types available all years
standard_offenses = [
    'Imprisonments for drug felonies',
    'Imprisonments for drug possession offenses',
    'Imprisonments for marijuana offenses',
    'Imprisonments for property felonies',
    'Imprisonments for violent felonies'
]

# Special handling for strike offenses
strike_offense_mapping = {
    'Imprisonments for strike offenses': [2009, 2010, 2011, 2012, 2014, 2015, 2016],
    'Imprisonments for 2nd strikes (June 30, 2013)': [2013],
    'Imprisonments for 3rd strikes (June 30, 2013)': [2013],
    'Imprisonments for 2nd strikes': [2012, 2014, 2015, 2016],  
    'Imprisonments for 3rd strikes': [2012, 2014, 2015, 2016]  
}

# Plot standard offenses
for offense in standard_offenses:
    yearly_data = []
    years = []
    
    for df in df_list_raw:
        year = df['Year'].iloc[0]
        if offense in df.columns:
            clean_values = pd.to_numeric(df[offense].replace('****', np.nan), errors='coerce')
            yearly_total = clean_values.sum()
            
            yearly_data.append(yearly_total)
            years.append(year)
    
    if yearly_data:
        sns.lineplot(
            x=years,
            y=yearly_data,
            label=offense.replace('Imprisonments for ', ''),
            marker='o',
            linewidth=2.5
        )

# Special handling for strike offenses
strike_data = {'2nd Strike': {}, '3rd Strike': {}}

for df in df_list_raw:
    year = df['Year'].iloc[0]
    
    # Find the right column name for this year
    for col in df.columns:
        if '2nd strike' in col.lower() or 'second strike' in col.lower():
            clean_values = pd.to_numeric(df[col].replace('****', np.nan), errors='coerce')
            strike_data['2nd Strike'][year] = clean_values.sum()
        elif '3rd strike' in col.lower() or 'third strike' in col.lower():
            clean_values = pd.to_numeric(df[col].replace('****', np.nan), errors='coerce')
            strike_data['3rd Strike'][year] = clean_values.sum()

# Plot strike offenses
for strike_type, year_values in strike_data.items():
    if year_values:
        years = sorted(year_values.keys())
        values = [year_values[y] for y in years]
        sns.lineplot(
            x=years,
            y=values,
            label=f'Strike Offenses ({strike_type})',
            marker='o',
            linewidth=2.5,
        )

plt.title('Total Imprisonments by Offense Type Across California Counties (2009-2016)')
plt.ylabel('Total Number of Imprisonments')
plt.xlabel('Year')
plt.legend(title='Offense Type', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Group offense types into violent vs non-violent
violent_offenses = [
    'Imprisonments for violent felonies',
]

non_violent_offenses = [
    'Imprisonments for drug felonies',
    'Imprisonments for drug possession offenses',
    'Imprisonments for marijuana offenses',
    'Imprisonments for property felonies'
]

plt.figure(figsize=(14, 8))

# Function to aggregate data by category
def aggregate_offenses(offense_list, df_list, label):
    years = []
    values = []
    
    for df in df_list:
        year = df['Year'].iloc[0]
        yearly_total = 0
        count = 0
        
        for offense in offense_list:
            if offense in df.columns:
                clean_values = pd.to_numeric(df[offense].replace('****', np.nan), errors='coerce')
                offense_sum = clean_values.sum()
                if not np.isnan(offense_sum):
                    yearly_total += offense_sum
                    count += 1
        
        if count > 0:  # Only add if we found data
            years.append(year)
            values.append(yearly_total)
    
    if years:
        sns.lineplot(
            x=years,
            y=values,
            label=label,
            marker='o'
        )

# Plot violent offenses
aggregate_offenses(violent_offenses, df_list_raw, 'Violent Offenses')

# Plot non-violent offenses
aggregate_offenses(non_violent_offenses, df_list_raw, 'Non-Violent Offenses')

plt.title('Imprisonments: Violent vs Non-Violent Offenses (2009-2016)')
plt.ylabel('Number of Imprisonments')
plt.xlabel('Year')
plt.legend(title='Offense Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Poverty Rate vs. Imprisonment Rate Over Time (Scatterplot)
poverty_columns = {
    2009: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 5-year average",
    2010: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 5-year average",
    2011: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 5-year average",
    2012: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 2010 Census",
    2013: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 2010 Census",
    2014: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 2010 Census",
    2015: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 2011-2015 average",
    2016: "Percent of adults ages 18-64 living in households with incomes below poverty guidelines, 2011-2016 average"
}

# Loop through each year's dataframe 
for df in df_list_measure:
    year = df['Year'].iloc[0]
    if year not in poverty_columns:
        continue

    poverty_col = poverty_columns[year]
    imprisonment_col = "Total adult imprisonments per 100,000/population age 18-69"

    if poverty_col in df.columns and imprisonment_col in df.columns:
        plt.figure(figsize=(10, 6))

        # Scatterplot with county coloring
        sns.scatterplot(
            data=df,
            x=poverty_col,
            y=imprisonment_col,
            hue="County",
            palette='tab20',
            alpha=0.7
        )

        # Regression line 
        sns.regplot(
            data=df,
            x=poverty_col,
            y=imprisonment_col,
            scatter=False,  
            color='black',  
            line_kws={"linewidth": 2}
        )

        plt.title(f"Poverty Rate vs Imprisonment Rate ({year})")
        plt.xlabel("Adult Poverty Rate (%)")
        plt.ylabel("Imprisonments per 100,000 Adults")
        plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1), ncol=1)
        plt.tight_layout()
        plt.show()
```

```python
# Strikes Over Time


strike_cols = [
    'Imprisonments for 2nd strikes per 100,000 population',
    'Imprisonments for 3rd strikes per 100,000 population'
]

# Check which strike columns are available
available_strike_cols = [col for col in strike_cols if col in measures_all_years.columns]

if available_strike_cols:
    plt.figure(figsize=(14, 8))
    for strike in available_strike_cols:
        sns.lineplot(
            data=measures_all_years,
            x='Year',
            y=strike,
            estimator='mean',
            label=strike,
            ci=None
        )
    plt.title('Average Strike Imprisonments Over Time (2009-2016)')
    plt.ylabel('Imprisonments per 100,000 Population')
    plt.xlabel('Year')
    plt.legend(title='Strike Type')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
# Imprisonments By Offense Type

offense_types = [
   'Imprisonments for 2nd strikes per 100,000 population',
   'Imprisonments for 3rd strikes per 100,000 population',
   'Imprisonments for drug offenses per 100,000 population',
   'Imprisonments for drug possession offenses per 100,000 population',
   'Imprisonments for marijuana offenses per 100,000 population', 
   'Imprisonments for property offenses per 100,000 population',
   'Imprisonments for strike offenses per 100,000 population',
   'Imprisonments for violent offenses per 100,000 population',
]

for year in sorted(measures_all_years['Year'].unique()):
    year_data = measures_all_years[measures_all_years['Year'] == year]
    
    available_offenses = [col for col in offense_types if col in year_data.columns]
    
    if available_offenses:
        offense_data = year_data[['County'] + available_offenses].set_index('County')
        
        plt.figure(figsize=(16, 10))
        ax = offense_data.plot(kind='bar', stacked=True, colormap='Set2')
        plt.title(f'Imprisonments by Offense Type per County ({year})', pad=20) 
        plt.ylabel('Imprisonments per 100,000 Population')
        plt.xlabel('County')
        plt.xticks(rotation=90)
        
        plt.legend(title='Offense Type', bbox_to_anchor=(1.05, 1), loc='upper left')
        
        plt.tight_layout(rect=[0, 0, 0.85, 1])  
        plt.show()
```
